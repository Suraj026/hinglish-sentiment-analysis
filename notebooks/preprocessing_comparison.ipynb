# Tokenizer comparison

Compare whitespace splitting with `Preprocessor`/NLTK TweetTokenizer on 20 training tweets.

```python
import os
os.environ.setdefault('OPENBLAS_NUM_THREADS', '1')
os.environ.setdefault('OMP_NUM_THREADS', '1')
os.environ.setdefault('MKL_NUM_THREADS', '1')

from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = next(path for path in [Path.cwd(), *Path.cwd().parents] if (path / 'src').is_dir())
sys.path.insert(0, str(PROJECT_ROOT))

from src.preprocess import PreprocessConfig, Preprocessor
from src.handoff import build_manifest, write_manifest

DATA_DIR = PROJECT_ROOT / 'data'
raw_train = pd.read_csv(DATA_DIR / 'train.tsv', sep='\t', header=None, names=['uid', 'text', 'label'])
tweets = raw_train['text'].fillna('').astype(str).tolist()

preprocessor = Preprocessor(PreprocessConfig()).fit(tweets)
comparison = [(tweet, tweet.split(), preprocessor.transform(tweet)) for tweet in tweets[:20]]

processed_dir = PROJECT_ROOT / 'data' / 'processed'
processed_dir.mkdir(parents=True, exist_ok=True)
preprocessor.save(processed_dir / 'preprocessor.json')

for filename, frame in [('train', raw_train), ('val', pd.read_csv(DATA_DIR / 'validation.tsv', sep='\t', header=None, names=['uid', 'text', 'label'])), ('test', pd.read_csv(DATA_DIR / 'test.tsv', sep='\t', header=None, names=['uid', 'text']))]:
    output = pd.DataFrame({'uid': frame['uid'], 'text_clean': [preprocessor.transform_text(text) for text in frame['text'].fillna('').astype(str)]})
    if 'label' in frame:
        output['label'] = frame['label'].astype(int)
    output.to_csv(processed_dir / f'{filename}.csv', index=False)

split_files = {name: processed_dir / f'{name}.csv' for name in ('train', 'val', 'test')}
write_manifest(build_manifest(split_files, 'H1', seed=42), processed_dir / 'H1_MANIFEST.json')
comparison
```
